🎯 O Seaborn é uma biblioteca de visualização em Python construída sobre o Matplotlib, que facilita a criação de gráficos estatísticos elegantes e informativos com poucas linhas de código.

Vamos ao passo a passo 👇
1. Importando bibliotecas

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

2. Conjunto de dados de exemplo

O Seaborn já traz datasets prontos para treino, como o famoso tips (gorjetas de restaurante):

```python
# Carregar dataset exemplo
df = sns.load_dataset("tips")
print(df.head())
```

3. Exemplos de gráficos Seaborn

📊 Gráfico de dispersão (scatterplot)

```python
sns.scatterplot(data=df, x="total_bill", y="tip", hue="sex", style="time")
plt.show()
```

📈 Gráfico de linha (lineplot)

```python
sns.lineplot(data=df, x="size", y="tip", estimator="mean")
plt.show()
```

📉 Gráfico de regressão (regplot)

```python
sns.regplot(data=df, x="total_bill", y="tip")
plt.show()
```

Gráfico de caixa (boxplot)

```python
sns.boxplot(data=df, x="day", y="total_bill", hue="sex")
plt.show()
```

🎻 Gráfico de violino (violinplot)

```python
sns.violinplot(data=df, x="day", y="tip", hue="sex", split=True)
plt.show()
```

🟦 Mapa de calor (heatmap)

```python
# Select only numeric columns for correlation calculation
numeric_df = df.select_dtypes(include=['number'])

corr = numeric_df.corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", linewidths=0.5)
plt.show()

```

4. Estilos e paletas

O Seaborn permite personalizar o estilo facilmente:

```python
sns.set_style("whitegrid")  # opções: white, dark, whitegrid, darkgrid, ticks
sns.set_palette("muted")    # paletas: deep, bright, pastel, dark, muted
```
